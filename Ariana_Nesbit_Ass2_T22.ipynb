```python
!pip install -U ultralytics
from ultralytics import YOLO
import torch
print('GPU available:', torch.cuda.is_available())
print('Device:', torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'CPU')
```

```python
from google.colab import drive
import shutil
import os

drive.mount('/content/drive')

DATASET_PATH = '/content/drive/MyDrive/kitti3k_yolo'
CAR_ONLY_PATH = '/content/drive/MyDrive/kitti3k_car_only'

shutil.copytree(DATASET_PATH, '/content/kitti3k_yolo')
print('Dataset copied to local storage.')
print(os.listdir('/content/kitti3k_yolo'))
```

```python
dataset_yaml = '/content/kitti3k_yolo/kitti3k.yaml'

## 1st CONFIGURATION
model_config1 = YOLO('yolo11n.pt')

train_results1 = model_config1.train(
    data=dataset_yaml,
    epochs=20,
    imgsz=640,
    batch=16,
    lr0=0.001,
    freeze=10,
    workers=2,
    cache='ram',
    device=0,
    exist_ok=True,
    pretrained=True
)

print('Configuration 1 training complete.')
```

```python
import shutil
shutil.copy('/content/runs/detect/train/weights/best.pt',
            '/content/drive/MyDrive/config1_best.pt')
print('Model saved to Drive.')
```

```python
## 2ND CONFIGURATION

model_config2 = YOLO('yolo11n.pt')

train_results2 = model_config2.train(
    data=dataset_yaml,
    epochs=20,
    imgsz=640,
    batch=16,
    lr0=0.001,
    freeze=10,
    workers=2,
    cache='ram',
    device=0,
    exist_ok=True,
    pretrained=True,
    # augmentation targeting small/rare objects like cyclists
    degrees=15,
    scale=0.8,
    mosaic=1.0,
    copy_paste=0.3,
    mixup=0.1,
    name='train2'
)

print('Configuration 2 training complete.')
```

```python
shutil.copy('/content/runs/detect/train2/weights/best.pt',
            '/content/drive/MyDrive/config2_best.pt')
print('Config 2 saved.')
```

```python
## 3RD CONFIGURATON

model_config3 = YOLO('yolo11n.pt')

train_results3 = model_config3.train(
    data=dataset_yaml,
    epochs=40,
    imgsz=640,
    batch=16,
    lr0=0.001,
    freeze=0,
    workers=2,
    cache='ram',
    device=0,
    exist_ok=True,
    pretrained=True,
    cos_lr=True,
    name='train3'
)

print('Configuration 3 training complete.')
```

```python
shutil.copy('/content/runs/detect/train3/weights/best.pt',
            '/content/drive/MyDrive/config3_best.pt')
print('Config 3 saved.')
```
